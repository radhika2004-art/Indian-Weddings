```python
theory_mean= 2+0.01*400
mean=6.5
std=20
sample=30
alpha=0.05
```

```python
import numpy as np
standard_error= std/np.sqrt(sample)
```

```python
t_statistics= (mean-theory_mean)/standard_error
```

```python
t_statistics
```

```python
from scipy.stats import norm
critical_value=norm.ppf(1-alpha)
critical_value
```

```python
if t_statistics > critical_value:
    decision = "H1"
else:
    decision = "H0"
```

```python
decision
```
