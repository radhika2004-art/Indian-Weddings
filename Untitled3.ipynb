```python
theory_mean= 1.8+0.02*450
mean=10.8
std=25
sample=35
alpha=0.05
```

```python
import numpy as np
standard_error= std/np.sqrt(sample)
```

```python
t_statistics= (mean-theory_mean)/standard_error
```

```python
t_statistics
```

```python
from scipy.stats import norm
critical_value=norm.ppf(1-alpha)
critical_value
```

```python
if t_statistics > critical_value:
    decision= "H0"
else:
    decision= "H1"
```

```python
decision
```
